```python
#count number of vowels in a given string
s = input()
vowels = "aeiouAEIOU"
count = 0
for i in s:
  if i in vowels:
    count+=1
print(count)

```

```python
#print reverse of the given number
def reverse(n):
  rev = 0
  while(n!=0):
    rem = n%10
    rev = rev*10 + rem
    n = n//10
  return rev
print(reverse(457))
```
