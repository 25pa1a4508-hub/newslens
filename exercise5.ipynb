```python
#program that print reverse of a string
def reverse(s):
  return s[::-1]
print(reverse("india"))
```

```python
#without using indexing and function
def reverse(str):
  for i in range(len(str)):
    print(str[len(str)-i-1],end="")
reverse("india")
```

```python
#functions that return multiple return values
def multiple_functions(*nums):
  s=sum(nums)
  c=len(nums)
  avg = s//c
  return s,c,avg
res_sum,res_count,res_avg=multiple_functions(1,2,3,4,5,7,8)
print(f"the sum is {res_sum}")
print(f"the count is {res_count}")
print(f"the average is {res_avg}")
```
